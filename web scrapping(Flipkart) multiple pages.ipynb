```python
# import libraries 
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
# creating empty list
product_name=[]
product_price=[]
product_desc=[]
product_review=[]
```

```python

for i in range(1,11):
    url='https://www.flipkart.com/search?q=mobile+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page='+str(i)
    r=requests.get(url)
    soup=BeautifulSoup(r.text,'lxml')
    box=soup.find('div',class_='DOjaWF gdgoEp')

    names=box.find_all('div',class_='KzDlHZ')
    for name in names:
        name=name.text
        product_name.append(name)
    #print(product_name)
    #print(len(product_name))

    prices=box.find_all('div',class_='Nx9bqj _4b5DiR')
    for price in prices:
        price=price.text
        product_price.append(price)
    #print(product_price)
    #print(len(product_price))

    desc=box.find_all('ul',class_='G4BRas')
    for des in desc:
        des=des.text
        product_desc.append(des)
    #print(product_desc)
    #print(len(product_desc))

    reviews=box.find_all('div',class_='XQDdHH')
    for review in reviews:
        review=review.text
        product_review.append(review)
    print(product_review)
    print(len(product_review))








```

```python
df=pd.DataFrame({
    'Product Name':product_name,
    'Product Price':product_price,
    'Product Description':product_desc,
    'Product Review':product_review
})
```
